```python
import pandas as pd, numpy as np

# 1) LOAD & CLEAN (all drivers, all laps from one weekend)
def hms_to_s(s):
    s=str(s).strip(); 
    if s in {"","-","nan"}: return np.nan
    p=s.split(":"); 
    return float(s) if len(p)==1 else (int(p[0])*60+float(p[1]) if len(p)==2 else int(p[0])*3600+int(p[1])*60+float(p[2]))

def as_bool(s):  # handles 'True'/'1'/etc.
    return pd.Series(s).astype(str).str.strip().str.lower().isin(["1","true","yes","y","t"])

def load_endurance(path, race_no):
    df = pd.read_csv(path, sep=";"); df.columns = df.columns.str.strip().str.upper()
    hms=lambda x: np.nan if pd.isna(x) or str(x).strip() in {"","-"} else sum(float(t)*m for t,m in zip(str(x).split(":")[::-1],[1,60,3600]))
    for c in ["LAP_TIME","S1","S2","S3","ELAPSED","HOUR","FL_TIME"]: 
        if c in df: df[c+"_S"]=df[c].map(hms)
    df["GREEN"]=df["FLAG_AT_FL"].astype(str).str.upper().eq("GF")
    s1 = as_bool(df.get("S1_LARGE", False)); s2 = as_bool(df.get("S2_LARGE", False)); s3 = as_bool(df.get("S3_LARGE", False))
    bad = s1 | s2 | s3
    df["LAP_TIME_S"] = df.get("LAP_TIME_S", np.nan)
    if df["LAP_TIME_S"].isna().any() and set(["S1_SECONDS","S2_SECONDS","S3_SECONDS"]).issubset(df.columns):
        df["LAP_TIME_S"] = df["LAP_TIME_S"].fillna(df[["S1_SECONDS","S2_SECONDS","S3_SECONDS"]].sum(axis=1))
    df["RACE_NO"]=race_no
    return df[df["GREEN"] & ~bad].copy()

laps = pd.concat([load_endurance(endurance_race_1_path, 1),
                  load_endurance(endurance_race_2_path, 2)], ignore_index=True)

# 2) TIDY FEATURES
laps["SESS_TIME_MIN"] = laps["ELAPSED_S"]/60     # for time-series plots
laps["AVG_KPH"] = laps["KPH"]
laps["CAR"] = laps["NUMBER"].astype(str)
laps["DRIVER"] = laps["DRIVER_NUMBER"].astype("Int64").astype(str)

# 3) PER-LAP (for ribbons/lines) & PER-DRIVER SUMMARY (for comparisons)
per_lap = laps[["RACE_NO","CAR","DRIVER","LAP_NUMBER","LAP_TIME_S","S1_SECONDS","S2_SECONDS","S3_SECONDS","AVG_KPH","SESS_TIME_MIN"]]

driver_summary = (laps.groupby(["RACE_NO","CAR","DRIVER"])
    .agg(laps_cnt=("LAP_NUMBER","count"),
         best_s=("LAP_TIME_S","min"),
         median_s=("LAP_TIME_S","median"),
         p95_s=("LAP_TIME_S", lambda x: x.quantile(0.95)),
         s1_med=("S1_SECONDS","median"),
         s2_med=("S2_SECONDS","median"),
         s3_med=("S3_SECONDS","median"),
         avg_kph=("AVG_KPH","mean"))
    .assign(consistency_s=lambda d: d["p95_s"]-d["median_s"])
    .reset_index())

race_summary = (driver_summary
    .assign(top3=lambda d: d.groupby(["RACE_NO"])["best_s"].rank(method="dense"))
    .sort_values(["RACE_NO","best_s"]))

print(per_lap.head(), "\n", driver_summary.head(), "\n", race_summary.head())
```

```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt

# --- START: plug your DataFrame here ---
summary = race_summary.copy()  # e.g., summary = driver_summary
# ---------------------------------------

# Normalize types/columns
summary["CAR"] = summary["CAR"].astype(str)
for c in ["best_s","median_s","p95_s","s1_med","s2_med","s3_med","avg_kph"]:
    if c in summary: summary[c] = pd.to_numeric(summary[c], errors="coerce")
if "consistency_s" not in summary.columns:
    summary["consistency_s"] = summary["p95_s"] - summary["median_s"]

# 1) Scatter: Best vs Median per race (labels = car no.)
for race in sorted(summary["RACE_NO"].unique()):
    sub = summary.query("RACE_NO==@race")
    plt.figure()
    plt.scatter(sub["best_s"], sub["median_s"])
    for _, r in sub.iterrows():
        plt.annotate(r["CAR"], (r["best_s"], r["median_s"]), xytext=(3,3),
                     textcoords="offset points", fontsize=8)
    plt.xlabel("Best lap (s)"); plt.ylabel("Median lap (s)")
    plt.title(f"Best vs Median — Race {race}")
    plt.show()

# 2) Bar: Consistency (p95 - median) sorted per race (lower = better)
for race in sorted(summary["RACE_NO"].unique()):
    sub = summary.query("RACE_NO==@race").sort_values("consistency_s")
    plt.figure()
    plt.bar(sub["CAR"], sub["consistency_s"])
    plt.xlabel("Car"); plt.ylabel("Consistency (s)")
    plt.title(f"Consistency by Car — Race {race}")
    plt.xticks(rotation=90); plt.tight_layout(); plt.show()

# 3) Stacked bars: sector medians (S1/S2/S3) per race
for race in sorted(summary["RACE_NO"].unique()):
    sub = summary.query("RACE_NO==@race").sort_values("median_s").copy()
    plt.figure()
    bottom = np.zeros(len(sub))
    for col in ["s1_med","s2_med","s3_med"]:
        plt.bar(sub["CAR"], sub[col], bottom=bottom, label=col.upper())
        bottom += sub[col].to_numpy()
    plt.xlabel("Car"); plt.ylabel("Median sector time (s)")
    plt.title(f"Sector Median Breakdown — Race {race}")
    plt.legend(); plt.xticks(rotation=90); plt.tight_layout(); plt.show()

# 4) Delta bars: R2 - R1 for cars present in both (best, median, consistency)
r1 = summary.query("RACE_NO==1").set_index("CAR")
r2 = summary.query("RACE_NO==2").set_index("CAR")
common = r1.join(r2, lsuffix="_r1", rsuffix="_r2", how="inner")
delta = pd.DataFrame({
    "CAR": common.index,
    "Δbest_s": common["best_s_r2"] - common["best_s_r1"],
    "Δmedian_s": common["median_s_r2"] - common["median_s_r1"],
    "Δconsistency_s": common["consistency_s_r2"] - common["consistency_s_r1"],
}).sort_values("Δmedian_s")

for col, title in [("Δbest_s","Best Lap Change (R2 - R1)"),
                   ("Δmedian_s","Median Lap Change (R2 - R1)"),
                   ("Δconsistency_s","Consistency Change (R2 - R1)")]:
    plt.figure()
    plt.bar(delta["CAR"], delta[col])
    plt.xlabel("Car"); plt.ylabel(col)
    plt.title(title); plt.xticks(rotation=90); plt.tight_layout(); plt.show()
```

```python
# Parse lap times to seconds

df = best_10_race_1.copy()

def t2s(x):
    m,s = str(x).split(':'); return int(m)*60+float(s)

best_cols  = [c for c in df.columns if c.startswith('BESTLAP_') and c.endswith(tuple(str(i) for i in range(1,11))) and '_LAPNUM' not in c]
avg_col    = 'AVERAGE' if 'AVERAGE' in df.columns else None

for c in best_cols + ([avg_col] if avg_col else []):
    df[c] = df[c].map(t2s)

# Pace features
df['best_of_best_s']   = df[best_cols].min(axis=1)                # outright speed
df['mean_top5_s']      = np.sort(df[best_cols].to_numpy(),axis=1)[:, :5].mean(axis=1)   # race-pace proxy
df['mean_top10_s']     = np.sort(df[best_cols].to_numpy(),axis=1)[:, :10].mean(axis=1)  # smoother race-pace
df['avg_reported_s']   = df[avg_col] if avg_col else np.nan

# Proxy “leaderboard”: first by laps completed, then by sustained pace
pace_board = (df.assign(TOTAL_DRIVER_LAPS=pd.to_numeric(df['TOTAL_DRIVER_LAPS'], errors='coerce'))
                .sort_values(['TOTAL_DRIVER_LAPS','mean_top10_s','best_of_best_s'],
                             ascending=[False, True, True])
                [['NUMBER','TOTAL_DRIVER_LAPS','best_of_best_s','mean_top5_s','mean_top10_s','avg_reported_s']])
print(pace_board.head(10))
```

```python
# ===== LEADERBOARD OVER TIME — LABEL EVERY CAR (start & end) =====
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from matplotlib.cm import get_cmap

# --- helpers ---
def to_seconds(x):
    if x is None: return np.nan
    s = str(x).strip()
    if not s or s.lower()=="nan": return np.nan
    if ":" in s:
        m, rest = s.split(":", 1)
        return float(m)*60 + float(rest)
    return float(s)

def _ensure_times(df):
    d = df.copy()
    if "LAP_s" not in d.columns: d["LAP_s"] = d["LAP_TIME"].map(to_seconds)
    d = d.dropna(subset=["NUMBER","LAP_NUMBER","LAP_s"])
    d["NUMBER"] = d["NUMBER"].astype(int)
    d["LAP_NUMBER"] = d["LAP_NUMBER"].astype(int)
    return d

def _compute_positions(d):
    d = d.sort_values(["NUMBER","LAP_NUMBER"]).copy()
    d["cum_s"] = d.groupby("NUMBER")["LAP_s"].cumsum()
    d = d.sort_values(["LAP_NUMBER","cum_s"])
    d["pos"] = d.groupby("LAP_NUMBER")["cum_s"].rank(method="min").astype(int)
    return d.sort_values(["NUMBER","LAP_NUMBER"]).reset_index(drop=True)

def _color_map(car_ids):
    # rotates through tab20 + tab20b + tab20c for bigger fields
    cmaps = [get_cmap("tab20"), get_cmap("tab20b"), get_cmap("tab20c")]
    colors = {}
    for i, car in enumerate(sorted(car_ids)):
        cmap = cmaps[(i // 20) % len(cmaps)]
        colors[car] = cmap(i % 20)
    return colors

def _spread(labels_y, min_gap=0.25, iters=3):
    y = np.array(labels_y, dtype=float)
    for _ in range(iters):
        order = np.argsort(y)
        for i in range(1, len(order)):
            a, b = order[i-1], order[i]
            if y[b] - y[a] < min_gap:
                y[b] = y[a] + min_gap
    return y

# --- main ---
def plot_field_leaderboard_all_labeled(
    df,
    fav_cars=None,                # e.g., [3, 55]
    show_mid_markers=True,        # put small markers every N laps
    every_n_laps=5,
    figsize=(13,7),
    lw_normal=1.4,
    lw_fav=2.8,
    alpha_line=0.85,
    alpha_others=0.55
):
    """
    Requires columns: NUMBER, LAP_NUMBER, LAP_TIME (M:SS.mmm or seconds)
    Labels EVERY car at start and finish. Optionally bold favorite cars.
    """
    fav_cars = set(fav_cars or [])
    d = _ensure_times(df)
    d = _compute_positions(d)

    last_lap = int(d["LAP_NUMBER"].max())
    first_lap = int(d["LAP_NUMBER"].min())
    cars = sorted(d["NUMBER"].unique())
    colors = _color_map(cars)

    # series per car
    lines = {car: grp[["LAP_NUMBER","pos"]].sort_values("LAP_NUMBER")
             for car, grp in d.groupby("NUMBER")}

    # collect start/end positions for all cars
    starts, ends = [], []
    for car, g in lines.items():
        starts.append((car, g["LAP_NUMBER"].iloc[0], g["pos"].iloc[0]))
        ends.append((car, g["LAP_NUMBER"].iloc[-1], g["pos"].iloc[-1]))

    # de-overlap labels
    start_y = _spread([p for (_,_,p) in starts], min_gap=0.25, iters=4)
    end_y   = _spread([p for (_,_,p) in ends],   min_gap=0.25, iters=4)

    # plot
    fig, ax = plt.subplots(figsize=figsize)

    # draw each car’s line
    for car, g in lines.items():
        is_fav = car in fav_cars
        ax.plot(
            g["LAP_NUMBER"], g["pos"],
            color=colors[car],
            linewidth=lw_fav if is_fav else lw_normal,
            alpha=alpha_line if is_fav else alpha_others
        )
        if show_mid_markers:
            sel = g[g["LAP_NUMBER"] % every_n_laps == 0]
            ax.scatter(sel["LAP_NUMBER"], sel["pos"], s=18,
                       color=colors[car], alpha=0.9 if is_fav else 0.6)

    # axis style
    ax.invert_yaxis()
    ax.set_xlabel("Lap")
    ax.set_ylabel("Position (lower is better)")
    ax.set_title("Leaderboard over time — every car labeled (start & finish)")
    ax.grid(True, linestyle="--", alpha=0.35)

    # start labels (left)
    x_left = first_lap - 0.5
    for (car, _, pos), ylab in zip(starts, start_y):
        ax.text(x_left, ylab, str(car),
                color=colors[car],
                va="center", ha="right",
                fontsize=9, fontweight="bold",
                bbox=dict(boxstyle="round,pad=0.15", facecolor="white", alpha=0.65, edgecolor="none"))
        # small connector
        x0 = first_lap
        y0 = lines[car]["pos"].iloc[0]
        ax.plot([x_left+0.1, x0-0.1], [ylab, y0], color=colors[car], linewidth=0.8, alpha=0.7)

    # end labels (right)
    x_right = last_lap + 0.5
    for (car, _, pos), ylab in zip(ends, end_y):
        ax.text(x_right, ylab, str(car),
                color=colors[car],
                va="center", ha="left",
                fontsize=9, fontweight="bold",
                bbox=dict(boxstyle="round,pad=0.15", facecolor="white", alpha=0.65, edgecolor="none"))
        # small connector
        x1 = last_lap
        y1 = lines[car]["pos"].iloc[-1]
        ax.plot([x1+0.1, x_right-0.1], [y1, ylab], color=colors[car], linewidth=0.8, alpha=0.7)

    # annotate favorites at the end with a simple callout
    if fav_cars:
        fav_txt = "Tracking: " + ", ".join(str(x) for x in sorted(fav_cars))
        ax.text(0.01, 1.01, fav_txt, transform=ax.transAxes, ha="left", va="bottom", fontsize=10)

    fig.tight_layout()
    plt.show()

# -------- Usage --------
plot_field_leaderboard_all_labeled(endurance_race_1, fav_cars=[3,72], show_mid_markers=True, every_n_laps=3)

```
