```python
provisional_race_2_path = "barber-motorsports-park/barber/03_Provisional Results_Race 2_Anonymized.CSV"
race_results_2 = pd.read_csv( provisional_race_2_path,header=0,sep=";")
```

```python
race_results_2['race'] = 2
race_results_2 = race_results_2[['POSITION', 'NUMBER','LAPS','FL_LAPNUM','FL_TIME','FL_KPH','TOTAL_TIME','GAP_FIRST','GAP_PREVIOUS','VEHICLE','TIRES','DIVISION','CLASS']]
race_results_2
```

```python

official_race_2_results_path = "barber-motorsports-park/barber/03_Results GR Cup Race 2 Official_Anonymized.CSV"
official_race_2_results = pd.read_csv(official_race_2_results_path, header=0,sep=";")
```

```python
official_race_2_results[['POSITION', 'NUMBER','LAPS','FL_LAPNUM','FL_TIME','FL_KPH','TOTAL_TIME','GAP_FIRST','GAP_PREVIOUS','VEHICLE','TIRES','DIVISION','CLASS']]
```

```python

A,B=provisional_race_2_path, official_race_2_results_path
raw_same = hashlib.sha256(open(A,'rb').read()).hexdigest()==hashlib.sha256(open(B,'rb').read()).hexdigest()
norm=lambda d: d.applymap(lambda x: str(x).strip() if isinstance(x,str) else x).reindex(sorted(d.columns),axis=1).sort_values(by=sorted(d.columns)).reset_index(drop=True)
sem_same = hashlib.sha256(norm(pd.read_csv(A, sep=None, engine='python')).to_csv(index=False).encode()).hexdigest()==hashlib.sha256(norm(pd.read_csv(B, sep=None, engine='python')).to_csv(index=False).encode()).hexdigest(); print({'raw_same':raw_same,'semantic_same':sem_same})
```
